Prereq: `rclone` configured with a remote named `gdrive` (see [`Data/external/README.md`](../Data/external/README.md)).

```
from pathlib import Path

from acquisition.drive import DriveRemote
from acquisition.run import run

REPO = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
MANIFEST = REPO / 'Data' / 'external' / 'manifest.yaml'
EXTERNAL = REPO / 'Data' / 'external'
DRIVE = DriveRemote(remote_name='gdrive', root='Lithium_v2/external')
```

## Run a single dataset

Useful for development. Pick `usgs`, `izquierdo`, `wetland2026`, or `spei`.

```
run(manifest_path=MANIFEST, external_root=EXTERNAL, drive=DRIVE, only={'usgs'})
```

## Run everything

Full pipeline. Idempotent: re-running with no upstream changes is a no-op apart from the `rclone lsf` checks.

```
run(manifest_path=MANIFEST, external_root=EXTERNAL, drive=DRIVE)
```
